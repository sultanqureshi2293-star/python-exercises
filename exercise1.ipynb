## Q1. String operations
Write a function that returns the reversed string and a string containing every second character.

For “every 2nd character”, this solution selects characters at indexes `0, 2, 4, ...` (the first, third, fifth, and subsequent alternating characters).


```python
def string_operations(text):
    reversed_text = text[::-1]
    every_second = text[::2]
    return reversed_text, every_second

sample = "Artificial Intelligence"
reversed_result, every_second_result = string_operations(sample)

print("Original string:", sample)
print("Reversed string:", reversed_result)
print("Every 2nd character:", every_second_result)
```

## Q2. Squares of even numbers using list comprehension
The condition keeps only even values, and each retained value is squared.


```python
def squares_of_even_numbers(numbers):
    return [number ** 2 for number in numbers if number % 2 == 0]

numbers = [1, 2, 3, 4, 5, 6, 8]
print("Original list:", numbers)
print("Squares of even numbers:", squares_of_even_numbers(numbers))
```

## Q3. Character frequency using a dictionary
The function counts how many times each character occurs. Spaces and letter case are counted as they appear in the input.


```python
def character_frequency(text):
    frequency = {}
    for character in text:
        frequency[character] = frequency.get(character, 0) + 1
    return frequency

sample = "hello"
print("Text:", sample)
print("Character frequency:", character_frequency(sample))
```

## Q4. Create a dictionary using zip()
`zip()` pairs items at the same positions. If the lists have different lengths, the resulting dictionary uses pairs up to the length of the shorter list.


```python
def lists_to_dictionary(keys, values):
    return dict(zip(keys, values))

subjects = ["Python", "C++", "Mathematics"]
marks = [90, 85, 88]
print(lists_to_dictionary(subjects, marks))
```

## Q5. Manual iteration using iter(), next(), and StopIteration
`iter()` creates an iterator. `next()` returns one item at a time. After the final item, `next()` raises `StopIteration`, which is handled by the `try-except` block.


```python
items = ["Python", "C++", "AI"]
iterator = iter(items)

try:
    while True:
        item = next(iterator)
        print(item)
except StopIteration:
    print("The iterator has no more elements.")
```

## Q6. Palindrome check using slicing
A palindrome reads the same forward and backward. The expression `text == text[::-1]` compares the original string with its reversed version. This implementation is case-sensitive and includes spaces.


```python
def is_palindrome(text):
    return text == text[::-1]

print("madam:", is_palindrome("madam"))
print("hello:", is_palindrome("hello"))
```

## Q7. Second largest number without sorting
The solution tracks the largest and second-largest **distinct** values in one pass. It does not sort the list. If fewer than two distinct values are present, it raises a clear `ValueError`.


```python
def second_largest(numbers):
    largest = None
    second = None

    for number in numbers:
        if largest is None or number > largest:
            second = largest
            largest = number
        elif number != largest and (second is None or number > second):
            second = number

    if second is None:
        raise ValueError("The list must contain at least two distinct numbers.")

    return second

values = [12, 35, 1, 10, 34, 35]
print("Numbers:", values)
print("Second largest distinct number:", second_largest(values))
```
